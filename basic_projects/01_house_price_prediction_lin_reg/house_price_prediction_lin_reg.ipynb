**Linear Regression**

```python
def lin_reg(x,w,b):
  m = x.shape[0]
  f_wb = np.zeros(m)
  for i in range(m):
    f_wb[i] = w * x[i] + b
  return f_wb
```

```python
w = 50
b = 190
```

```python
apply_lin_reg = lin_reg(x_train, y_train, w, b)
print(apply_lin_reg[:10])
```

**Compute Cost using Mean Squared Error**

```python
# J(w,b) = 1/2m * (f_wb - y^i)^2

def compute_cost(x , y, w, b):
  m = x.shape[0]
  total_cost = 0
  for i in range(m):
    f_wb = w * x[i] + b
    cost = (f_wb - y[i]) ** 2
    total_cost += cost
  total_cost = total_cost/(2 * m)
  return total_cost
```
