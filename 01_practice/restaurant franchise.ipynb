```python
# Compute cost

def compute_cost(x,y,w,b):
  total_cost = 0
  m = x.shape[0]
  for i in range(m):
    f_wb = np.dot(w, x[i]) + b
    cost = (f_wb - y[i]) ** 2
    total_cost = total_cost + cost
  total_cost = total_cost/(2 * m)
  return total_cost
```

```python
# Compute Gradient to calculate the derivative w.r.t w and b

def compute_gradient(x,y,w,b):
  dj_dw = 0
  dj_db = 0
  m = x.shape[0]
  for i in range(m):
    f_wb = np.dot(w , x[i]) + b
    error = (f_wb - y[i])
    dj_dw_i = error * x[i]
    dj_db_i = error

    dj_dw = dj_dw + dj_dw_i
    dj_db = dj_db + dj_db_i

  dj_dw = dj_dw / m
  dj_db = dj_db / m

  return dj_dw , dj_db
```

```python
dw , db = compute_gradient(x_train, y_train , w_in , b_in)
print(dw)
print(db)
```

```python
# Gradient descent
def gradient_descent(x, y, w, b, alpha, iters, cost_fun, compute_gradient_fun):

  cost_hist = []
  wb_history = []
  w = copy.deepcopy(w_in)
  b = b_in

  for i in range(iters):
    dj_dw, dj_db = compute_gradient_fun(x,y,w,b)

    w = w - alpha * dj_dw
    b = b - alpha * dj_dw

    if i > 10000:
      cost_hist.append(cost_fun(x,y,w,b))

    if i % math.ceil(iters/10) == 10:
      wb_history.append(w)
      print(f"Iterations: {i : 4}: Cost: {cost_hist[-1]:8.2f}")

    return w, b, cost_hist, wb_history
```
