```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])

print(x_train)
print(y_train)

print(x_train.shape)
print(y_train.shape)

print(x_train.ndim)
print(y_train.ndim)
```

```python
def plot_data(x,y,ax):
  pos = (y == 1)
  neg = (y == 0)

  ax.scatter(x[pos , 0] ,
             x[pos , 1] ,
             label = "y = 1" ,
             marker = 'x' ,
             c= 'r')

  ax.scatter(x[neg , 0] ,
             x[neg , 1] ,
             label = "y = 0" ,
             marker = 'o' ,
             c= 'b')

  ax.set_title("Logisitc Regression")
  ax.legend()

```

```python
fig, ax = plt.subplots(1,1, figsize=(4,4))
plot_data(x_train, y_train, ax)
ax.axis([0,4,0,3.5])
plt.show()
```

```python
def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
# Logistic cost function

def logistic_cost(x, y, w, b):
  m = x.shape[0]
  cost = 0

  for i in range(m):
    z_i = np.dot(x[i] , w) + b
    # first we need to calculate the loss for each example. we will later accumulate all the loss in the cost varaible.
    # we first need sigmoid
    f_wb_i = sigmoid(z_i)
    loss = - y[i] * np.log(f_wb_i) - (1 - y[i]) * np.log(1 - f_wb_i)
    cost = cost + loss


  cost = cost / m
  return cost
```

```python
w = np.array([1, 1])
b = -3
```

```python
cost = logistic_cost(x_train, y_train, w, b)
print(cost)
```

```python
x0_line = np.arange(0, 6)
x1_b3 = 3 - x0_line
x1_b4 = 4 - x0_line
fig, ax = plt.subplots(1,1, figsize=(4,4))
ax.plot(x0_line, x1_b3 , c= 'black', label="$b=-3$")
ax.plot(x0_line, x1_b4, label = "$b=-4$" , c='magenta')
ax.axis([0,4,0,4])

plot_data(x_train, y_train, ax)
ax.axis([0,4,0,4])
ax.set_title("Decision Boundary")
ax.set_xlabel("$x_0$")
ax.set_ylabel("$x_1$")
```
