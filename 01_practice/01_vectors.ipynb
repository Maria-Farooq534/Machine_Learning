Scalar Vector Operations

```python
a = np.array([1,2,3,4])
b = 5 * a

print(a)
print(b)
```

```python
# Dot product
a = np.array([1,2,3,4])
b = np.array([1,2,5,7])

def dot_product(x, y):
  dot_p = 0
  m = x.shape[0]
  for i in range(m):
    xy_i = x[i] * y[i]
    dot_p += xy_i
  return dot_p


cal_dot = dot_product(a,b)
print(cal_dot)
```

```python
# using numpy dot() function
a = np.array([1,-2,3,4])
b = np.array([1,2,5,7])

c = np.dot(a,b)
print(c)
print(c.shape) # it will show () bcz, its a scalar

d = np.dot(b,a)
print(d)
```

```python
import time
```

```python
print("%%%%% Loop Version %%%%%")
print(" ")

a = np.array([1,-2,3,4])
b = np.array([1,2,5,7])
tic = time.time()
loop_results = dot_product(a,b)
print(f"Dot product result using loop: {loop_results}")
toc = time.time()
print(f"Start time: {tic:.4f} , Stop Time: {toc:.4f}")
time_stamp = toc-tic
print(f"Loop version duration: {1000 * time_stamp:.4f}ms")

print(" ")
print("%%%%% dot() Function Version %%%%%")
print(" ")

# now lets do the dot product by numpy dot function
tic = time.time()
fun_result = np.dot(a,b)
print(f"Dot Product Computeed by np.dot() function: {fun_result}")
toc = time.time()
print(f"Start Time before np.dot(): {tic:.4f} , End Time: {toc:.4f}")
print(f"np.dot() function Duration: {1000 * (toc-tic):.4f}ms")
```

```python
a = np.arange(6).reshape(-1, 2)       # the reshape(-1,2) means, there should be 2 columns, and the remaining will be rows, range of 6: 2 * 3 = 6 ,so 2 rows and 3 columns
print(a)
print(a.shape)


```

```python
# indexing in matrix

print(a[2,1])  # it returns the scalar, in index 2, the row is 3,  [4 5], then column index 1, the column is 2, so , the targeted value is 5.
print(a[2,1].shape) # will return () , coz , its a scalar
print(a[2])   # now we have not specified the clumn, we only provide 1 value , that is for row, so index 2 is row 3. it will give [4,5]
print(a[2].shape) # now its 1D array, it will give (2,)
```
