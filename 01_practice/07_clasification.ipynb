```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])
```

```python
X_train2 = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train2 = np.array([0, 0, 0, 1, 1, 1])


print(X_train)
print(y_train)

print(X_train.ndim)
print(y_train.ndim)


print(X_train.shape)
print(y_train.shape)
```

```python
pos = (y_train == 1)
neg = (y_train == 0)

print(pos)
print(neg)
```

```python
print(X_train[pos])
print(y_train[pos])
```

```python
print(X_train[pos].shape)
print(y_train[pos].shape)

print(pos.shape)
print(neg.shape)
```

```python
pos2 = X_train2 == 1
neg2 = y_train == 0
```

```python
fig, ax = plt.subplots(1, 2 , figsize=(12, 3))
ax[0].scatter(X_train[pos] , y_train[pos] , marker = 'x' , label = 'y=1', s=80 , c='r')
ax[0].scatter(X_train[neg] , y_train[neg] , marker = 'o' , c='b' , label = 'y=0' , s=100)



ax[0].set_xlabel("X")
ax[0].set_ylabel("Y")
ax[0].set_title("One variable plot")
ax[0].legend()


ax[1].scatter(X_train2[pos2, 0], X_train2[pos2 , 1], marker='x' , s=80 , label = "y=1")
ax[1].scatter(X_train2[neg2, 0], X_train2[neg2, 1], marker='o' , label = 'y=0' , s=100)
ax[1].set_xlabel("x0")
ax[1].set_ylabel("x1")
ax[1].set_title("Two variable plot")
ax[1].legend()


plt.legend()
plt.show()
```
