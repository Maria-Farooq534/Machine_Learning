```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
X_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0, 0, 0, 1, 1, 1])

w_in = np.zeros(1)
b_in = 0
```

```python
print(np.c_[X_train,y_train])
```

```python
z = w_in * X_train + b_in
print(z)
```

```python
def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
y_hat = sigmoid(z)
print(y_hat)
```

```python
def logistic_cost(y,y_hat):
  cost = -1 * np.log(y_hat) - (1-y) * np.log(1-y_hat)
  return cost
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1]).reshape(-1, 1)
```

```python
print(X_train)
print(y_train)

print(X_train.shape)
print(y_train.shape)

print(X_train.ndim)
print(y_train.ndim)
```

```python
print(y_train ==1)
```

```python
positive = (y_train ==1).flatten()
print(positive)
print(positive.shape)
```

```python
print(X_train[positive])
```

```python
negative = (y_train == 0).flatten()
print(negative)
print(negative.shape)
print(X_train[negative])
```

```python
print(positive)
print(type(positive))
print(positive.dtype)
```

```python
print(X_train[positive, 0])
print(X_train[positive, 1])
```

```python
def plot_data(x,y,ax):
  positive = (y == 1).flatten()
  negative = (y == 0).flatten()

  ax.scatter(x[positive , 0], x[positive, 1], marker = 'x' , label = 'y = 1')
  ax.scatter(x[negative , 0], x[negative, 1], marker = 'o' , label = 'y = 0')
```

```python
fig, ax = plt.subplots(1, 1, figsize=(4,4))

plot_data(X_train, y_train, ax)
ax.axis([0, 4, 0, 3.5])
ax.set_xlabel("x_0")
ax.set_ylabel("x_1")
ax.legend()
plt.show()
```

```python
z = np.arange(-10 , 11)
print(z)
```

```python
def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
z_sigmoid = sigmoid(z)
print(z_sigmoid)
```

```python
x0 = np.arange(0, 6)
x1 = 3 - x0
print(x0)
print(x1)
```

```python
fig, ax = plt.subplots(1,1, figsize=(5,3))
ax.plot(x0, x1, c='b')
ax.fill_between(x0, x1, alpha=0.2)
ax.axis([0, 4, 0, 3.5])
plot_data(X_train, y_train, ax)
ax.set_ylabel(r'$x_1$')
ax.set_xlabel(r'$x_0$')
plt.show()
```
