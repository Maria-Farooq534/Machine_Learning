```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
X_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0, 0, 0, 1, 1, 1])

w_in = np.zeros(1)
b_in = 0
```

```python
print(np.c_[X_train,y_train])
```

```python
z = w_in * X_train + b_in
print(z)
```

```python
def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
y_hat = sigmoid(z)
print(y_hat)
```

so now, the actual predicition is y_train = [0 0 0 1 1 1] , but model is predicting [0.5 0.5 0.5 0.5 0.5 0.5] for all.
so, it means model is not predicting correctly and has not learned anything yet.

```python
def logistic_cost(y,y_hat):
  cost = -1 * np.log(y_hat) - (1-y) * np.log(1-y_hat)
  return cost
```

```python
w_in = np.array([1.])
b_in = -2.5
```

```python
z = w_in * X_train + b_in
print(z)
```

```python
y_hat = sigmoid(z)
print(y_hat)
```

```python
logisticcost = logistic_cost(y_train , y_hat)
print(logisticcost)
```
