```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([2, 3])
print(f"X: {x}")
print(f"Shape of X: {x.shape}")
```

```python
# creating parameters

W1 = np.array([[0.5, 1], [1, -1]])
b1 = np.array([0, 1])

print(f"W1 is: {W1}")
print(f"b1 is: {b1}")

print(W1.shape)
print(b1.shape)

print(W1.dtype)
print(b1.dtype)
```

```python
z1 = np.dot(x, W1) + b1
print(z1)
print(z1.shape)
print(type(z1))
print(z1.ndim)
print(z1.dtype)
```

```python
z1 = np.dot(x, W1) + b1
print(z1)
print(z1.shape)
```

```python
z1 = np.dot(W1, x) + b1
print(z1)
print(z1.shape)
```

```python
def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
a1 = sigmoid(z1)
print(a1)
print(a1.shape)
print(a1.ndim)
print(type(a1))
```

```python
W2 = np.array([[1, 2],[-1, 1]])
b2 = np.array([0, 0.5])
print(W2)
print(b2)
print(a1)
```

```python
z2 = np.dot(W2, a1) + b2
print(z2)
print(z2.shape)
print(z2.ndim)

a2 = sigmoid(z2)
print(a2)
print(a2.shape)
```

```python
# For output layer

W3 = np.array([2, -1])
b3 = np.array([-0.5])

z3 = np.dot(W3, a2) + b3
a3 = sigmoid(z3)
print(z3)
print(a3)
print(a3.dtype)
print(type(a3))
```

```python
# apply threshold
prediction = 1 if a3 >= 0.5 else 0
print(f"The output class is: {prediction}")
```

```python
# all layers are using same method, so i can create a function and then reuse it when needed.

def dense(W, a_in, b):
  z = np.dot(W, a_in) + b
  a_out = sigmoid(z)
  return z, a_out
```

```python
z1 , a1 = dense(W1, x, b1)
print(f"z1: {z1} , a1: {a1}")

z2, a2 = dense(W2, a1, b2)
print(f"z2: {z2} , a2: {a2}")

z3, a3 = dense(W3, a2, b3)
print(f"z3: {z3} , a3: {a3}")
```
