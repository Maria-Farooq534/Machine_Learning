```python
# Sigmoid

def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
z_tmp = np.arange(-10,11)
print(z_tmp)
```

```python
y = sigmoid(z_tmp)
print(y)

np.set_printoptions(precision=3)
print(np.c_[z_tmp , y])
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12,3))
ax.plot(z_tmp , y , c = 'b')
ax.set_xlabel("z")
ax.set_ylabel("sigmoid(z)")
ax.set_title("Sigmoid Function")
```
